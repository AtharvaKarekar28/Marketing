```python
import sys, os
sys.path.append("..")

from dotenv import load_dotenv
from openai import OpenAI
import pandas as pd

from src.personas import generate_personas
from src.simulator import ask_persona

load_dotenv("../.env")
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))
MODEL = "gpt-4.1-mini"
```

```python
scenario = "Netflix is raising the price of its standard plan by $3 per month."
personas = generate_personas(5)

results = [ask_persona(client, MODEL, p, scenario) for p in personas]
pd.DataFrame(results)
```

```python
for r in results:
    print(f"[{r['decision']}] {r['reason']}\n")
```
